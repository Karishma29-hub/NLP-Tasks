```python
from collections import defaultdict

class FiveGramModel:

    def __init__(self):
        self.fivegrams = defaultdict(lambda: defaultdict(int))

    def train(self, corpus):
        for sentence in corpus:
            words = sentence.lower().split()

            if len(words) < 5:
                continue

            for i in range(len(words)-4):
                context = tuple(words[i:i+4])
                next_word = words[i+4]
                self.fivegrams[context][next_word] += 1

    def predict(self, text):

        words = text.lower().split()

        if len(words) < 4:
            return []

        context = tuple(words[-4:])

        if context not in self.fivegrams:
            return []

        prediction = sorted(
            self.fivegrams[context].items(),
            key=lambda x: x[1],
            reverse=True
        )

        return [word for word, count in prediction]


corpus = [
    "best places to visit in india",
    "best places to visit in chennai",
    "best places to visit near me",
    "best places to visit during summer",
    "best places to eat in chennai",
]

model = FiveGramModel()
model.train(corpus)

query = "best places to visit"

print(model.predict(query))
```
