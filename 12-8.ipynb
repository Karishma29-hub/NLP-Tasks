```python
from textblob.en import spelling

print("Spelling dictionary loaded!")
print("Number of words:", len(spelling))
```

```python
vocabulary = set(spelling.keys())

print("Vocabulary size:", len(vocabulary))
```

```python
def edit_distance(word1, word2):

    m = len(word1)
    n = len(word2)

    dp = [[0] * (n + 1) for _ in range(m + 1)]

    for i in range(m + 1):
        dp[i][0] = i

    for j in range(n + 1):
        dp[0][j] = j

    for i in range(1, m + 1):
        for j in range(1, n + 1):

            if word1[i - 1] == word2[j - 1]:
                cost = 0
            else:
                cost = 1

            dp[i][j] = min(
                dp[i - 1][j] + 1,
                dp[i][j - 1] + 1,
                dp[i - 1][j - 1] + cost
            )

    return dp[m][n]
```

```python
print("Edit Distance:", edit_distance("machne", "machine"))
```

```python
def get_closest_word(word):

    best_word = word
    best_distance = float("inf")

    for candidate in vocabulary:

        distance = edit_distance(word, candidate)

        if distance < best_distance:
            best_distance = distance
            best_word = candidate

    return best_word, best_distance
```

```python
import re

def correct_query(query):

    words = query.lower().split()

    corrected_words = []
    incorrect_words = []
    suggestions = []

    for word in words:

        clean_word = re.sub(r"[^a-zA-Z]", "", word)

        if clean_word in vocabulary:

            corrected_words.append(clean_word)

        else:

            incorrect_words.append(clean_word)

            suggested_word, distance = get_closest_word(clean_word)

            suggestions.append(
                (clean_word, suggested_word, distance)
            )

            corrected_words.append(suggested_word)

    corrected_query = " ".join(corrected_words)

    return incorrect_words, suggestions, corrected_query
```

```python
query = input("Enter search query: ")

incorrect_words, suggestions, corrected_query = correct_query(query)
```

```python
print("\n" + "=" * 60)
print("SPELLING CORRECTION RESULT")
print("=" * 60)

print("\nOriginal Query:")
print(query)

print("\nIncorrect Words:")

if incorrect_words:
    for word in incorrect_words:
        print("-", word)
else:
    print("No spelling errors found.")

print("\nSuggested Corrections:")

if suggestions:
    for wrong, correct, distance in suggestions:
        print(f"{wrong} -> {correct} (Edit Distance: {distance})")
else:
    print("No corrections required.")

print("\nFinal Corrected Query:")
print(corrected_query)

print("=" * 60)
```
